# Gráficos Descriptivos

Este script crea visualizaciones descriptivas del dataset de Pokémon usando `pandas`, `matplotlib` y `seaborn`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="muted")

# Cargar datos

df = pd.read_excel("data_sets/Pokemon.xlsx")

df.head()
```

```python
# Gráfico 1: Histograma de HP, Attack, Defense y Speed
numeric_cols = ["HP", "Attack", "Defense", "Speed"]

plt.figure(figsize=(12, 8))
for i, column in enumerate(numeric_cols, 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[column].dropna(), kde=True, color="royalblue")
    plt.title(f"Distribución de {column}")
    plt.xlabel(column)
    plt.ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
# Gráfico 2: Boxplot de estadísticas base por tipo principal
if "Type 1" in df.columns:
    plt.figure(figsize=(14, 6))
    sns.boxplot(x="Type 1", y="Total", data=df, order=df.groupby("Type 1")["Total"].median().sort_values(ascending=False).index)
    plt.xticks(rotation=45, ha="right")
    plt.title("Distribución de Total por Tipo Principal")
    plt.xlabel("Tipo 1")
    plt.ylabel("Total")
    plt.tight_layout()
    plt.show()
```

```python
# Gráfico 3: Conteo de Pokémon Legendary vs no Legendary
if "Legendary" in df.columns:
    plt.figure(figsize=(6, 4))
    sns.countplot(x="Legendary", data=df, palette=["#4c72b0", "#dd8452"])
    plt.title("Conteo de Pokémon Legendary vs No Legendary")
    plt.xlabel("Legendary")
    plt.ylabel("Cantidad")
    plt.tight_layout()
    plt.show()
```

```python
# Gráfico 4: Scatter plot Attack vs Defense, color por Tipo 1 si está disponible
plt.figure(figsize=(10, 8))
if "Type 1" in df.columns:
    sns.scatterplot(x="Attack", y="Defense", hue="Type 1", data=df, palette="tab10", edgecolor="w", alpha=0.8)
    plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", borderaxespad=0.)
else:
    sns.scatterplot(x="Attack", y="Defense", data=df, color="#2a9d8f", edgecolor="w", alpha=0.8)

plt.title("Attack vs Defense")
plt.xlabel("Attack")
plt.ylabel("Defense")
plt.tight_layout()
plt.show()
```
