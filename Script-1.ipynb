```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid", palette="muted")
```

```python
df = pd.read_excel("data_sets/Pokemon.xlsx")
df.head()
```

# Análisis Descriptivo

A continuación se realiza un análisis descriptivo del conjunto de datos de Pokémon para entender su estructura, distribución y características principales.

```python
# Información general del DataFrame
print("Dimensiones del dataset:", df.shape)
print("\nPrimeras filas:")
print(df.head())
print("\nTipos de datos y valores nulos:")
print(df.info())

# Estadísticas descriptivas para variables numéricas
print("\nResumen estadístico de variables numéricas:")
print(df.describe())

# Conteo de valores únicos para columnas categóricas importantes
categorical_columns = [col for col in df.columns if df[col].dtype == 'object']
print("\nColumnas categóricas relevantes:", categorical_columns)
for col in categorical_columns:
    print(f"\nValores únicos en '{col}' ({df[col].nunique()}):")
    print(df[col].value_counts().head(10))

# Análisis adicional de algunas columnas clave
if 'Type 1' in df.columns:
    print("\nTop 10 tipos principales de Pokémon:")
    print(df['Type 1'].value_counts().head(10))

if 'Legendary' in df.columns:
    print("\nDistribución Legendary vs no Legendary:")
    print(df['Legendary'].value_counts())
```

# Gráficos Descriptivos

En esta sección se generan visualizaciones para explorar la distribución de estadísticas y la relación entre atributos clave.

```python
# Histogramas de estadísticas principales
numeric_cols = ["HP", "Attack", "Defense", "Speed"]
plt.figure(figsize=(12, 8))
for i, column in enumerate(numeric_cols, 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[column].dropna(), kde=True, color="royalblue")
    plt.title(f"Distribución de {column}")
    plt.xlabel(column)
    plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()
```

```python
# Boxplot de Total por Type 1
if "Type 1" in df.columns:
    plt.figure(figsize=(14, 6))
    sns.boxplot(x="Type 1", y="Total", data=df, order=df.groupby("Type 1")["Total"].median().sort_values(ascending=False).index)
    plt.xticks(rotation=45, ha="right")
    plt.title("Distribución de Total por Tipo Principal")
    plt.xlabel("Tipo 1")
    plt.ylabel("Total")
    plt.tight_layout()
    plt.show()
```

```python
# Conteo Legendary vs no Legendary
if "Legendary" in df.columns:
    plt.figure(figsize=(6, 4))
    sns.countplot(x="Legendary", data=df, palette=["#4c72b0", "#dd8452"])
    plt.title("Conteo de Pokémon Legendary vs No Legendary")
    plt.xlabel("Legendary")
    plt.ylabel("Cantidad")
    plt.tight_layout()
    plt.show()
```

```python
# Scatter plot Attack vs Defense
plt.figure(figsize=(10, 8))
if "Type 1" in df.columns:
    sns.scatterplot(x="Attack", y="Defense", hue="Type 1", data=df, palette="tab10", edgecolor="w", alpha=0.8)
    plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", borderaxespad=0.)
else:
    sns.scatterplot(x="Attack", y="Defense", data=df, color="#2a9d8f", edgecolor="w", alpha=0.8)
plt.title("Attack vs Defense")
plt.xlabel("Attack")
plt.ylabel("Defense")
plt.tight_layout()
plt.show()
```
